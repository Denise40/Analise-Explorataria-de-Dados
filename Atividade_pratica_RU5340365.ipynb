```python
#Importando as bibliotecas necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
!pip install bokeh_sampledata
```

```python
# Abrir o dataframe
from bokeh.sampledata.iris import flowers as data
```

```python
# Criando uma cópia do dataframe
df = data.copy()

# Exibindo as primeiras cinco linhas
print("PRIMEIRAS CINCO LINHAS DO DATAFRAME:")
display(df.head())
```

```python
# Exibindo as dimensões do dataframe
print("\nDIMENSÕES DO DATAFRAME:")
print("Quantidade de linhas e colunas:", df.shape)
print("Total de linhas:", df.shape[0])
print("Total de colunas:", df.shape[1])
```

```python
# Exibindo os tipos de dados
print("\nTIPOS DE DADOS DAS COLUNAS:")
print(df.dtypes)
```

```python
# Exibindo os nomes das colunas
print("\nNOMES DAS COLUNAS:")
print(df.columns.tolist())
```

## 2. RESUMO DAS PRINCIPAIS ESTATÍSTICAS DO DATASET

```python
print("RESUMO ESTATÍSTICO DO DATASET:")
display(df.describe())
```

##3. FREQUÊNCIA ABSOLUTA DAS ESPÉCIES

```python
print("FREQUÊNCIA ABSOLUTA DAS ESPÉCIES:")

frequencia_absoluta = df["species"].value_counts()
display(frequencia_absoluta)

```

## 4. FREQUÊNCIA RELATIVA DAS ESPÉCIES

```python
print("FREQUÊNCIA RELATIVA DAS ESPÉCIES:")

frequencia_relativa = (df["species"].value_counts(normalize=True) * 100)

display(frequencia_relativa.round(2))
```

## 5. GRÁFICO DE BARRAS VERTICAIS - TOTAL POR ESPÉCIE

```python
plt.figure(figsize=(9, 5))

frequencia_absoluta.plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Quantidade de Flores por Espécie", fontsize=14)
plt.xlabel("Espécies")
plt.ylabel("Quantidade")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

## 6. GRÁFICO DE BARRAS HORIZONTAIS - TOTAL POR ESPÉCIE

```python
plt.figure(figsize=(9, 5))

frequencia_absoluta.plot(kind="barh", color="skyblue", edgecolor="black")

plt.title("Quantidade de Flores por Espécie", fontsize=14)
plt.xlabel("Quantidade")
plt.ylabel("Espécies")
plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

## 7. BOXPLOT DO DATASET - TODAS AS VARIÁVEIS POR ESPÉCIE

```python
plt.figure(figsize=(12, 8))

# Boxplot do comprimento da sépala
plt.subplot(2, 2, 1)
plt.boxplot([
    df[df["species"] == "setosa"]["sepal_length"],
    df[df["species"] == "versicolor"]["sepal_length"],
    df[df["species"] == "virginica"]["sepal_length"]
])
plt.title("Comprimento da Sépala")
plt.xticks([1, 2, 3], ["Setosa", "Versicolor", "Virginica"])

# Boxplot da largura da sépala
plt.subplot(2, 2, 2)
plt.boxplot([
    df[df["species"] == "setosa"]["sepal_width"],
    df[df["species"] == "versicolor"]["sepal_width"],
    df[df["species"] == "virginica"]["sepal_width"]
])
plt.title("Largura da Sépala")
plt.xticks([1, 2, 3], ["Setosa", "Versicolor", "Virginica"])

# Boxplot do comprimento da pétala
plt.subplot(2, 2, 3)
plt.boxplot([
    df[df["species"] == "setosa"]["petal_length"],
    df[df["species"] == "versicolor"]["petal_length"],
    df[df["species"] == "virginica"]["petal_length"]
])
plt.title("Comprimento da Pétala")
plt.xticks([1, 2, 3], ["Setosa", "Versicolor", "Virginica"])

# Boxplot da largura da pétala
plt.subplot(2, 2, 4)
plt.boxplot([
    df[df["species"] == "setosa"]["petal_width"],
    df[df["species"] == "versicolor"]["petal_width"],
    df[df["species"] == "virginica"]["petal_width"]
])
plt.title("Largura da Pétala")
plt.xticks([1, 2, 3], ["Setosa", "Versicolor", "Virginica"])

plt.tight_layout()
plt.show()
```

## 8. CÁLCULO DA MÉDIA E MEDIANA AGRUPADAS POR ESPÉCIE

```python
print("MÉDIA DAS VARIÁVEIS POR ESPÉCIE:")

variaveis_numericas = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

media = df.groupby("species")[variaveis_numericas].mean()

display(media.round(3))

print("\nMEDIANA DAS VARIÁVEIS POR ESPÉCIE:")

mediana = df.groupby("species")[variaveis_numericas].median()

display(mediana.round(3))
```

## 9. CÁLCULO DO DESVIO PADRÃO E DA VARIÂNCIA AGRUPADOS POR ESPÉCIE

```python
print("DESVIO PADRÃO POR ESPÉCIE:")

desvio_padrao = df.groupby("species")[variaveis_numericas].std()

display(desvio_padrao.round(3))

print("\nVARIÂNCIA POR ESPÉCIE:")

variancia = df.groupby("species")[variaveis_numericas].var()

display(variancia.round(3))
```
